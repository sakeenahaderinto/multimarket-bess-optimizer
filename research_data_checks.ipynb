```python
PRE2023_RENAME = {
    'Service': 'service',
    'Delivery Start': 'delivery_start',
    'Delivery End': 'delivery_end',
    'Cleared Volume': 'cleared_volume',
    'Clearing Price': 'clearing_price',
}

FY_RENAME = {
    'auctionProduct': 'service',
    'deliveryStart': 'delivery_start',
    'deliveryEnd': 'delivery_end',
    'clearedVolume': 'cleared_volume',
    'clearingPrice': 'clearing_price',
}

def process_dc_data(pre2023_results_summary, fy2023, fy2024, fy2025):

    # Filter dataset
    pre2023_filtered = pre2023_results_summary[pre2023_results_summary['Service'].isin(['DCL', 'DCH'])].copy()
    fy2023_filtered  = fy2023[fy2023['auctionProduct'].isin(['DCL', 'DCH'])].copy()
    fy2024_filtered  = fy2024[fy2024['auctionProduct'].isin(['DCL', 'DCH'])].copy()
    fy2025_filtered  = fy2025[fy2025['auctionProduct'].isin(['DCL', 'DCH'])].copy()

    # drop redundant columns (fully captured by delivery_start/delivery_end)
    pre2023_filtered = pre2023_filtered.drop(columns=['EFA', 'EFA Date'])

    # rename columns to canonical names — note 'service' matches pipeline.py's
    # dc_df["service"] reference exactly, so this combined output is a drop-in
    # replacement for the old single-source dc_auction_results.parquet
    pre2023_filtered = pre2023_filtered.rename(columns=PRE2023_RENAME)
    fy2023_filtered  = fy2023_filtered.rename(columns=FY_RENAME)
    fy2024_filtered  = fy2024_filtered.rename(columns=FY_RENAME)
    fy2025_filtered  = fy2025_filtered.rename(columns=FY_RENAME)

    # parse datetimes before concat, so sort_values works correctly
    for df in (pre2023_filtered, fy2023_filtered, fy2024_filtered, fy2025_filtered):
        df['delivery_start'] = pd.to_datetime(df['delivery_start'])
        df['delivery_end']   = pd.to_datetime(df['delivery_end'])

    # concatenate — long format, one row per service per delivery window
    # note: auctionID / linkedServiceWindowID only exist on FY20XX sources —
    # pre-2023 rows will be NaN for these by design (no native equivalent)
    combined = pd.concat(
        [pre2023_filtered, fy2023_filtered, fy2024_filtered, fy2025_filtered],
        ignore_index=True,
    )
    combined = combined.sort_values('delivery_start').reset_index(drop=True)

    return combined
```

```python
def check_source_boundaries(combined: pd.DataFrame) -> pd.DataFrame:
    """
    Checks for gaps or overlaps at the boundaries between consecutive rows
    in a combined DC dataset, based on delivery_start / delivery_end.

    DCL and DCH are parallel series (both can report the same delivery window),
    so contiguity is checked independently within each service_type rather than
    across the combined frame — otherwise interleaved DCL/DCH rows at the same
    timestamp would be flagged as false overlaps.

    Inputs:
        combined: DataFrame from process_dc_data(), with columns service_type,
                  delivery_start, delivery_end.
    Outputs:
        DataFrame of only the rows where a gap or overlap was found, with
        columns: service_type, delivery_end, next_delivery_start, diff.
        Empty DataFrame means no issues found.
    """
    all_issues = []

    for service, group in combined.groupby('service'):
        df = group.sort_values('delivery_start').reset_index(drop=True)
        next_start = df['delivery_start'].shift(-1)
        diff = next_start - df['delivery_end']

        issues = pd.DataFrame({
            'service': service,
            'delivery_end': df['delivery_end'],
            'next_delivery_start': next_start,
            'diff': diff,
        })
        # diff == 0 means perfectly adjacent — expected, not an issue
        issues = issues[issues['diff'] != pd.Timedelta(0)].dropna(subset=['diff'])
        all_issues.append(issues)

    issues = pd.concat(all_issues, ignore_index=True) if all_issues else pd.DataFrame()

    if issues.empty:
        print("No gaps or overlaps found — all rows are contiguous within each service.")
    else:
        n_gaps = (issues['diff'] > pd.Timedelta(0)).sum()
        n_overlaps = (issues['diff'] < pd.Timedelta(0)).sum()
        print(f"Found {n_gaps} gap(s) and {n_overlaps} overlap(s).")

    return issues
```

```python
DA_RENAME = {
    'Delivery_Start_UTC': 'datetime',
    'IMRP_Amount_GBP_per_MWh': 'value',
}

def process_day_ahead_data(day_ahead: pd.DataFrame) -> pd.DataFrame:
    """
    Standardizes the raw NESO day-ahead (IMRP) data to the schema pipeline.py
    expects: a 'datetime' column (UTC) and a 'value' column (price, GBP/MWh).

    Source data is hourly (Settlement_Period 1-24, or 23/25 on clock-change
    days). pipeline.py's resample('30min').ffill() step upsamples this to
    half-hourly automatically — no resampling needed here.

    Inputs:
        day_ahead: raw NESO IMRP DataFrame with columns IMRP_Date,
                   Settlement_Period, Delivery_Start_Local, Delivery_Start_UTC,
                   IMRP_Amount_GBP_per_MWh.
    Outputs:
        DataFrame with columns: datetime, value. Sorted by datetime.
    """
    df = day_ahead[['Delivery_Start_UTC', 'IMRP_Amount_GBP_per_MWh']].copy()
    df = df.rename(columns=DA_RENAME)

    df['datetime'] = pd.to_datetime(df['datetime'], utc=True)
    df = df.sort_values('datetime').reset_index(drop=True)

    return df
```
